# **Uvod**

U ovoj svesci vrši se učitavanje prethodno istreniranog i sačuvanog klasifikacionog modela pomoću Joblib biblioteke.

Korisniku je omogućeno interaktivno testiranje modela unosom naziva proizvoda, nakon čega model predviđa njegovu kategoriju. Testiranje se može ponavljati unosom novih proizvoda, a izvršavanje programa završava se unosom komande exit.

# **Uvoz biblioteke i učitavanje modela**

Uvozimo `joblib` biblioteku koja omogućava čuvanje i učitavanje prethodno istreniranih Python modela. Učitavamo sačuvani model iz .pkl fajla i smeštamo ga u promenljivu.

```python
import joblib

# Load the saved model
model = joblib.load("product_classifier_final_model.pkl")
```

# **Priprema iteraktivnog testiranja**

Prikazujemo korisniku instrukcije:
- Da unese naziv proizvoda koji želi da klasifikuje.
- Da unese exit ukoliko želi da prekine testiranje.



```python
# Interactive testing
print("\nEnter a product title to predict its category.")
print("Type 'exit' to stop.\n")
```

# **Predikcija kategorije proizvoda**


Kreiramo while petlju koja omogućava korisniku da unosi nazive proizvoda jedan za drugim.

Pomoću funkcije input() preuzimamo uneti tekst i čuvamo ga u promenljivoj title.

Zatim proveravamo da li je korisnik uneo reč exit. Ukoliko jeste, prekidamo izvršavanje programa pomoću naredbe break. U suprotnom, prosleđujemo uneti naziv proizvoda metodi model.predict(), koja pomoću prethodno istreniranog modela predviđa njegovu kategoriju.

Na kraju štampamo predviđenu kategoriju unetog proizvoda.

```python
while True:
    title = input("Enter product title: ")
    if title.strip().lower() == "exit":
            print("Exiting...")
            break

    prediction = model.predict([title])[0]
    print(f"Predicted category: {prediction}\n" + "-" * 40)
```

# **Zaključak**

U ovoj svesci uspešno smo učitali prethodno istrenirani model i omogućili njegovu praktičnu primenu za klasifikaciju novih proizvoda.

Kroz interaktivni unos testirali smo sposobnost modela da na osnovu naziva proizvoda predvidi odgovarajuću kategoriju, bez potrebe za ponovnim treniranjem.

Na ovaj način prikazali smo kako se prethodno istrenirani model može sačuvati, ponovo učitati i koristiti za buduće predikcije.
